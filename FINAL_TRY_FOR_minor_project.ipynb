```python

import pandas as pd
import re
import os

BASE = "/content/drive/MyDrive/TeluguShield_AI"
TRAIN_FILE = os.path.join(
    BASE, "01_Datasets/Original",
    "training_data_telugu-hate.xlsx"
)
TEST_FILE = os.path.join(
    BASE, "01_Datasets/Original",
    "telugu-english-test-data-with-labels.xlsx"
)

train = pd.read_excel(TRAIN_FILE)
test = pd.read_excel(TEST_FILE)

def normalize(text):
    text = str(text).lower().strip()
    text = re.sub(r"\s+", " ", text)
    return text

train["normalized"] = train["Comments"].apply(normalize)
test["normalized"] = test["Comments"].apply(normalize)

# 1. Duplicate comments inside training
train_duplicates = train[
    train.duplicated("normalized", keep=False)
].sort_values("normalized")

# 2. Check conflicting labels in training
train_conflicts = (
    train.groupby("normalized")["Label"]
    .nunique()
)
train_conflicts = train_conflicts[train_conflicts > 1]

# 3. Duplicate comments inside test
test_duplicates = test[
    test.duplicated("normalized", keep=False)
].sort_values("normalized")

# 4. Overlap between training and test
overlap = set(train["normalized"]) & set(test["normalized"])

overlap_review = pd.concat([
    train[train["normalized"].isin(overlap)].assign(Source="Training"),
    test[test["normalized"].isin(overlap)].assign(Source="Test")
]).sort_values("normalized")

# 5. Cross-dataset label conflicts
cross_conflicts = []

for comment in overlap:
    train_labels = set(
        train.loc[train["normalized"] == comment, "Label"]
    )
    test_labels = set(
        test.loc[test["normalized"] == comment, "Label"]
    )

    if train_labels != test_labels:
        cross_conflicts.append({
            "Comment": comment,
            "Training_Labels": str(train_labels),
            "Test_Labels": str(test_labels)
        })

cross_conflicts = pd.DataFrame(cross_conflicts)

print("Training duplicate rows:", len(train_duplicates))
print("Training conflicting comment groups:", len(train_conflicts))
print("Test duplicate rows:", len(test_duplicates))
print("Cross-dataset overlapping unique comments:", len(overlap))
print("Cross-dataset label conflicts:", len(cross_conflicts))

print("\nTRAINING CONFLICTS:")
display(
    train_duplicates[
        train_duplicates["normalized"].isin(train_conflicts.index)
    ][["Comments", "Label"]]
)

print("\nCROSS-DATASET CONFLICTS:")
display(cross_conflicts)

# Save audit report
review_path = os.path.join(
    BASE, "02_Preprocessing",
    "duplicate_conflict_audit.xlsx"
)

with pd.ExcelWriter(review_path) as writer:
    train_duplicates.to_excel(
        writer, sheet_name="Training_Duplicates", index=False
    )
    test_duplicates.to_excel(
        writer, sheet_name="Test_Duplicates", index=False
    )
    overlap_review.to_excel(
        writer, sheet_name="Train_Test_Overlap", index=False
    )
    cross_conflicts.to_excel(
        writer, sheet_name="Cross_Conflicts", index=False
    )

print("\nAudit report saved:", review_path)
```

```python

import os
import pandas as pd

BASE = "/content/drive/MyDrive/TeluguShield_AI"

TRAIN_FILE = os.path.join(
    BASE, "01_Datasets/Original",
    "training_data_telugu-hate.xlsx"
)

TEST_FILE = os.path.join(
    BASE, "01_Datasets/Original",
    "telugu-english-test-data-with-labels.xlsx"
)

train = pd.read_excel(TRAIN_FILE)
test = pd.read_excel(TEST_FILE)

def normalize(text):
    return " ".join(str(text).lower().split())

train["normalized"] = train["Comments"].apply(normalize)
test["normalized"] = test["Comments"].apply(normalize)

# Identify conflicting training comments
conflict_groups = (
    train.groupby("normalized")["Label"].nunique()
)

conflict_keys = conflict_groups[
    conflict_groups > 1
].index

training_review = train[
    train["normalized"].isin(conflict_keys)
][["S.No", "Comments", "Label"]].copy()

# Identify cross-dataset conflict
cross_overlap = set(train["normalized"]) & set(test["normalized"])

cross_review = []

for comment in cross_overlap:
    train_labels = set(
        train.loc[train["normalized"] == comment, "Label"]
    )

    test_labels = set(
        test.loc[test["normalized"] == comment, "Label"]
    )

    if train_labels != test_labels:
        cross_review.append({
            "Comment": comment,
            "Training_Label": ", ".join(train_labels),
            "Test_Label": ", ".join(test_labels),
            "Final_Decision": "",
            "Reason": ""
        })

cross_review = pd.DataFrame(cross_review)

# Save review workbook
output = os.path.join(
    BASE, "02_Preprocessing",
    "Manual_Label_Review.xlsx"
)

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    training_review.to_excel(
        writer,
        sheet_name="Training_Conflicts",
        index=False
    )

    cross_review.to_excel(
        writer,
        sheet_name="Cross_Dataset_Conflict",
        index=False
    )

print("Manual review workbook created!")
print("Training conflict rows:", len(training_review))
print("Cross-dataset conflicts:", len(cross_review))
print("Saved at:", output)
```

```python

import os
import pandas as pd

base = "/content/drive/MyDrive/TeluguShield_AI"
original = base + "/01_Datasets/Original"
processed = base + "/01_Datasets/Processed"
review_path = base + "/02_Preprocessing/Manual_Label_Review.xlsx"

os.makedirs(processed, exist_ok=True)

# Load original datasets and reviewed workbook
train = pd.read_excel(original + "/training_data_telugu-hate.xlsx")
test = pd.read_excel(original + "/telugu-english-test-data-with-labels.xlsx")
review = pd.read_excel(review_path, sheet_name="Training_Conflicts")

def normalize(text):
    return " ".join(str(text).lower().split())

for df in [train, test, review]:
    df["normalized"] = df["Comments" if df is not test else "Comments"].apply(normalize)

# Build reviewed-label mapping
review["normalized"] = review["Comments"].apply(normalize)
review["Final_Label"] = review["Final_Label"].astype(str).str.strip().str.lower()

decisions = {}
for comment, group in review.groupby("normalized"):
    labels = group["Final_Label"].unique().tolist()
    if len(labels) != 1 or labels[0] not in ["hate", "non-hate", "ambiguous"]:
        raise ValueError(f"Invalid review decision: {comment}")
    decisions[comment] = labels[0]

# Check test duplicate labels before removing duplicates
test["normalized"] = test["Comments"].apply(normalize)

test_conflicts = test.groupby("normalized")["Label"].nunique()
if (test_conflicts > 1).any():
    raise ValueError("Test dataset has duplicate comments with conflicting labels. Stop for review.")

# Check every conflicting training comment was reviewed
train["normalized"] = train["Comments"].apply(normalize)
original_conflicts = train.groupby("normalized")["Label"].nunique()
unreviewed = [
    c for c in original_conflicts[original_conflicts > 1].index
    if c not in decisions
]
if unreviewed:
    raise ValueError(f"Unreviewed training conflicts: {unreviewed}")

# Apply reviewed labels; preserve original Label column
train["Final_Label"] = train.apply(
    lambda r: decisions.get(r["normalized"], str(r["Label"]).strip().lower()),
    axis=1
)

# Exclude ambiguous comments
ambiguous_removed = (train["Final_Label"] == "ambiguous").sum()
train = train[train["Final_Label"] != "ambiguous"].copy()

# Remove training comments that overlap with test
test_comments = set(test["normalized"])
overlap = train["normalized"].isin(test_comments)
overlap_removed = overlap.sum()
train = train[~overlap].copy()

# Verify remaining duplicate training labels are consistent
conflicts_left = train.groupby("normalized")["Final_Label"].nunique()
if (conflicts_left > 1).any():
    raise ValueError("Unresolved duplicate-label conflicts remain in training.")

# Deduplicate
train_before = len(train)
train = train.drop_duplicates("normalized").copy()
train_duplicates_removed = train_before - len(train)

test_before = len(test)
test = test.drop_duplicates("normalized").copy()
test_duplicates_removed = test_before - len(test)

# Remove helper column
train.drop(columns=["normalized"], inplace=True)
test.drop(columns=["normalized"], inplace=True)

# Save separately; original files remain unchanged
train.to_excel(processed + "/telugu_hate_train_cleaned.xlsx", index=False)
test.to_excel(processed + "/telugu_hate_test_cleaned.xlsx", index=False)

train.to_csv(processed + "/telugu_hate_train_cleaned.csv", index=False)
test.to_csv(processed + "/telugu_hate_test_cleaned.csv", index=False)

print("CLEANING COMPLETED")
print("Ambiguous rows removed:", ambiguous_removed)
print("Train-test overlap rows removed:", overlap_removed)
print("Training duplicates removed:", train_duplicates_removed)
print("Final training rows:", len(train))
print("Test duplicates removed:", test_duplicates_removed)
print("Final test rows:", len(test))

print("\nTraining labels:")
print(train["Final_Label"].value_counts())

print("\nTest labels:")
print(test["Label"].value_counts())

print("\nFiles saved in:", processed)
```

```python

# ============================================================
# TELUGUSHIELD AI - MuRIL DATA EFFICIENCY EXPERIMENT
# 25%, 50%, 75%, 100% - ALL FOUR IN ONE RUN
# ============================================================

!pip install -q transformers datasets accelerate scikit-learn sentencepiece

import os, gc, json, random, shutil
import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    EarlyStoppingCallback,
    DataCollatorWithPadding
)
from torch.utils.data import Dataset

# ---------------- CONFIGURATION ----------------

SEED = 42
MODEL_NAME = "google/muril-base-cased"
MAX_LENGTH = 128
EPOCHS = 5
BATCH_SIZE = 8
LEARNING_RATE = 2e-5

BASE = "/content/drive/MyDrive/TeluguShield_AI"
DATA_PATH = BASE + "/01_Datasets/Processed/telugu_hate_train_cleaned.xlsx"
SAVE_DIR = BASE + "/03_Baseline_Experiments/MuRIL"

os.makedirs(SAVE_DIR, exist_ok=True)

# Reproducibility
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

if not torch.cuda.is_available():
    raise RuntimeError("GPU not detected. Enable T4 GPU in Colab.")

print("GPU:", torch.cuda.get_device_name(0))

# ---------------- LOAD DATA ----------------

df = pd.read_excel(DATA_PATH)

df["Final_Label"] = df["Final_Label"].str.strip().str.lower()
df = df[df["Final_Label"].isin(["hate", "non-hate"])].copy()

label_map = {"non-hate": 0, "hate": 1}
df["label"] = df["Final_Label"].map(label_map)

print("\nDataset:", len(df))
print(df["Final_Label"].value_counts())

# Same fixed validation split for all experiments
train_pool, val_df = train_test_split(
    df,
    test_size=0.20,
    random_state=SEED,
    stratify=df["label"]
)

train_pool = train_pool.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)

# Nested stratified subsets: 25%, 50%, 75%, 100%
subsets = {}

for percentage in [25, 50, 75, 100]:
    if percentage == 100:
        subset = train_pool.copy()
    else:
        subset, _ = train_test_split(
            train_pool,
            train_size=percentage / 100,
            random_state=SEED,
            stratify=train_pool["label"]
        )

    subsets[percentage] = subset.reset_index(drop=True)

print("\nTraining subset sizes:")
for p, data in subsets.items():
    print(f"{p}%:", len(data))

print("Validation size:", len(val_df))

# ---------------- TOKENIZER ----------------

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

class TeluguDataset(Dataset):
    def __init__(self, dataframe):
        self.texts = dataframe["Comments"].astype(str).tolist()
        self.labels = dataframe["label"].astype(int).tolist()

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        item = tokenizer(
            self.texts[idx],
            truncation=True,
            max_length=MAX_LENGTH
        )
        item["labels"] = self.labels[idx]
        return item

val_dataset = TeluguDataset(val_df)
collator = DataCollatorWithPadding(tokenizer=tokenizer)

# ---------------- METRICS ----------------

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=1)

    precision, recall, macro_f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="macro",
        zero_division=0
    )

    return {
        "accuracy": accuracy_score(labels, predictions),
        "precision": precision,
        "recall": recall,
        "macro_f1": macro_f1
    }

# ---------------- FOUR EXPERIMENTS ----------------

results = []

for percentage, subset in subsets.items():

    print("\n" + "=" * 60)
    print(f"STARTING MuRIL - {percentage}% EXPERIMENT")
    print("Training examples:", len(subset))
    print("=" * 60)

    run_dir = os.path.join(SAVE_DIR, f"MuRIL_{percentage}_percent")
    os.makedirs(run_dir, exist_ok=True)

    train_dataset = TeluguDataset(subset)

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=2,
        id2label={0: "non-hate", 1: "hate"},
        label2id={"non-hate": 0, "hate": 1}
    )

    args = TrainingArguments(
        output_dir=run_dir + "/checkpoints",
        num_train_epochs=EPOCHS,
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=16,
        learning_rate=LEARNING_RATE,
        weight_decay=0.01,
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="eval_macro_f1",
        greater_is_better=True,
        save_total_limit=1,
        fp16=True,
        logging_strategy="epoch",
        report_to="none",
        seed=SEED,
        data_seed=SEED,
        dataloader_num_workers=2
    )

    trainer = Trainer(
        model=model,
        args=args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        data_collator=collator,
        compute_metrics=compute_metrics,
        callbacks=[
            EarlyStoppingCallback(early_stopping_patience=2)
        ]
    )

    trainer.train()

    # Best model is restored automatically
    metrics = trainer.evaluate()

    # Save only the best model and tokenizer
    best_dir = run_dir + "/best_model"
    trainer.save_model(best_dir)
    tokenizer.save_pretrained(best_dir)

    result = {
        "model": "MuRIL",
        "data_percentage": percentage,
        "training_samples": len(subset),
        "validation_samples": len(val_df),
        "accuracy": metrics["eval_accuracy"],
        "precision": metrics["eval_precision"],
        "recall": metrics["eval_recall"],
        "macro_f1": metrics["eval_macro_f1"],
        "best_epoch": trainer.state.best_model_checkpoint
    }

    results.append(result)

    with open(run_dir + "/metrics.json", "w") as f:
        json.dump(result, f, indent=4)

    # Save progress after every experiment
    pd.DataFrame(results).to_csv(
        SAVE_DIR + "/MuRIL_progress.csv",
        index=False
    )

    print("\nEXPERIMENT COMPLETED:", percentage, "%")
    print("Accuracy:", result["accuracy"])
    print("Macro-F1:", result["macro_f1"])

    del trainer, model, train_dataset
    gc.collect()
    torch.cuda.empty_cache()

# ---------------- FINAL SUMMARY ----------------

results_df = pd.DataFrame(results)

results_df.to_csv(SAVE_DIR + "/MuRIL_final_results.csv", index=False)
results_df.to_excel(SAVE_DIR + "/MuRIL_final_results.xlsx", index=False)

print("\n" + "=" * 65)
print("ALL FOUR MuRIL EXPERIMENTS COMPLETED")
print("=" * 65)

display(results_df)

print("\nResults saved at:", SAVE_DIR)
```

```python

# ============================================================
# TELUGUSHIELD AI - COMPLETE BASELINE EXPERIMENT AUDIT
# 3 MODELS x 4 DATA PERCENTAGES = 12 EXPERIMENTS
# ============================================================

import os
import json
import pandas as pd
import numpy as np

from pathlib import Path

# -------------------- 1. CONFIGURATION --------------------

BASE_DIR = Path(
    "/content/drive/MyDrive/TeluguShield_AI/"
    "03_Baseline_Experiments"
)

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/TeluguShield_AI/"
    "08_Results_and_Graphs"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

MODELS = {
    "MuRIL": "MuRIL",
    "TeluguBERT": "TeluguBERT",
    "XLM-RoBERTa": "XLM-RoBERTa"
}

PERCENTAGES = [25, 50, 75, 100]

summary_rows = []
epoch_rows = []
file_rows = []

# -------------------- 2. HELPER FUNCTIONS --------------------

def find_files(folder, filenames):
    found = []

    if not folder.exists():
        return found

    for root, dirs, files in os.walk(folder):
        for filename in filenames:
            if filename in files:
                found.append(Path(root) / filename)

    return found


def load_json(path):
    try:
        with open(path, "r") as f:
            return json.load(f)
    except Exception as e:
        print("Could not read:", path, "|", str(e))
        return None


def extract_epoch_from_checkpoint(checkpoint):
    if not checkpoint:
        return None

    try:
        return int(Path(checkpoint).name.split("-")[-1])
    except Exception:
        return None


# -------------------- 3. SCAN ALL EXPERIMENTS --------------------

print("=" * 65)
print("TELUGUSHIELD AI - EXPERIMENT AUDIT")
print("=" * 65)

for model_name, folder_name in MODELS.items():

    model_dir = BASE_DIR / folder_name

    print("\nMODEL:", model_name)

    for percentage in PERCENTAGES:

        # Experiment folder names
        if model_name == "XLM-RoBERTa":
            run_name = f"XLMR_{percentage}_percent"
        else:
            run_name = f"{model_name}_{percentage}_percent"

        run_dir = model_dir / run_name
        best_dir = run_dir / "best_model"

        print("\nChecking:", run_name)

        # -------------------- FILE AUDIT --------------------

        expected_files = [
            "config.json",
            "tokenizer_config.json",
            "special_tokens_map.json",
            "model.safetensors",
            "pytorch_model.bin"
        ]

        existing_model_files = []

        if best_dir.exists():
            for filename in expected_files:
                if (best_dir / filename).exists():
                    existing_model_files.append(filename)

        model_weights_found = (
            "model.safetensors" in existing_model_files
            or "pytorch_model.bin" in existing_model_files
        )

        model_config_found = (
            "config.json" in existing_model_files
        )

        model_saved = (
            best_dir.exists()
            and model_weights_found
            and model_config_found
        )

        file_rows.append({
            "model": model_name,
            "data_percentage": percentage,
            "experiment_folder_exists": run_dir.exists(),
            "best_model_folder_exists": best_dir.exists(),
            "model_weights_found": model_weights_found,
            "model_config_found": model_config_found,
            "existing_files": ", ".join(existing_model_files)
        })

        # -------------------- METRICS JSON --------------------

        metrics_path = run_dir / "metrics.json"

        metrics = load_json(metrics_path) if metrics_path.exists() else {}

        if not isinstance(metrics, dict):
            metrics = {}

        # -------------------- FIND TRAINER STATE --------------------

        state_files = find_files(
            run_dir,
            ["trainer_state.json"]
        )

        trainer_state = None
        state_path = None

        if state_files:
            state_path = state_files[-1]
            trainer_state = load_json(state_path)

        # -------------------- EXTRACT TRAINING HISTORY --------------------

        history = []

        if isinstance(trainer_state, dict):
            history = trainer_state.get("log_history", [])

        for entry in history:

            if not isinstance(entry, dict):
                continue

            epoch_rows.append({
                "model": model_name,
                "data_percentage": percentage,
                "epoch": entry.get("epoch"),
                "step": entry.get("step"),
                "training_loss": entry.get("loss"),
                "validation_loss": entry.get("eval_loss"),
                "validation_accuracy": entry.get("eval_accuracy"),
                "validation_macro_f1": entry.get("eval_macro_f1"),
                "learning_rate": entry.get("learning_rate")
            })

        # -------------------- BEST EPOCH --------------------

        best_checkpoint = None
        best_epoch = None
        best_metric = None

        if isinstance(trainer_state, dict):

            best_checkpoint = trainer_state.get(
                "best_model_checkpoint"
            )

            best_metric = trainer_state.get("best_metric")

            best_step = trainer_state.get(
                "best_global_step"
            )

            if best_step is not None:

                matching_entries = [
                    x for x in history
                    if x.get("step") == best_step
                    and "eval_macro_f1" in x
                ]

                if matching_entries:
                    best_epoch = matching_entries[-1].get("epoch")

            if best_epoch is None and best_checkpoint:
                checkpoint_name = Path(best_checkpoint).name

                checkpoint_matches = [
                    x for x in history
                    if x.get("step") == extract_epoch_from_checkpoint(
                        best_checkpoint
                    )
                    and "eval_macro_f1" in x
                ]

                if checkpoint_matches:
                    best_epoch = checkpoint_matches[-1].get("epoch")

        # -------------------- METRICS --------------------

        accuracy = metrics.get(
            "accuracy",
            metrics.get("eval_accuracy")
        )

        precision = metrics.get(
            "precision",
            metrics.get("eval_precision")
        )

        recall = metrics.get(
            "recall",
            metrics.get("eval_recall")
        )

        macro_f1 = metrics.get(
            "macro_f1",
            metrics.get("eval_macro_f1")
        )

        # -------------------- LOSS ANALYSIS --------------------

        train_losses = [
            x["loss"] for x in history
            if x.get("loss") is not None
        ]

        validation_losses = [
            x["eval_loss"] for x in history
            if x.get("eval_loss") is not None
        ]

        best_f1_values = [
            x["eval_macro_f1"] for x in history
            if x.get("eval_macro_f1") is not None
        ]

        final_train_loss = (
            train_losses[-1] if train_losses else None
        )

        final_validation_loss = (
            validation_losses[-1] if validation_losses else None
        )

        # Matched epoch loss gap, if available
        loss_gap = None

        if history:
            train_epoch_entries = [
                x for x in history
                if x.get("loss") is not None
                and x.get("epoch") is not None
            ]

            eval_epoch_entries = [
                x for x in history
                if x.get("eval_loss") is not None
                and x.get("epoch") is not None
            ]

            if train_epoch_entries and eval_epoch_entries:

                train_entry = train_epoch_entries[-1]
                eval_entry = min(
                    eval_epoch_entries,
                    key=lambda x: abs(
                        x["epoch"] - train_entry["epoch"]
                    )
                )

                if abs(
                    train_entry["epoch"] - eval_entry["epoch"]
                ) < 0.01:

                    loss_gap = (
                        eval_entry["eval_loss"]
                        - train_entry["loss"]
                    )

        # -------------------- AUDIT STATUS --------------------

        if not run_dir.exists():
            status = "EXPERIMENT FOLDER MISSING"

        elif not metrics_path.exists():
            status = "METRICS JSON MISSING"

        elif not model_saved:
            status = "MODEL FILES INCOMPLETE"

        elif not state_files:
            status = "MODEL SAVED - TRAINING LOGS MISSING"

        else:
            status = "MODEL AND TRAINING LOGS FOUND"

        # -------------------- SUMMARY ROW --------------------

        summary_rows.append({

            "model": model_name,

            "data_percentage": percentage,

            "training_samples": metrics.get(
                "training_samples"
            ),

            "validation_samples": metrics.get(
                "validation_samples"
            ),

            "accuracy": accuracy,

            "precision": precision,

            "recall": recall,

            "macro_f1": macro_f1,

            "best_epoch": best_epoch,

            "best_metric_from_trainer": best_metric,

            "best_checkpoint": best_checkpoint,

            "last_training_loss": final_train_loss,

            "last_validation_loss": final_validation_loss,

            "validation_training_loss_gap": loss_gap,

            "epochs_recorded": len(best_f1_values),

            "trainer_state_found": bool(state_files),

            "model_saved": model_saved,

            "audit_status": status
        })

        print("Status:", status)
        print("Macro-F1:", macro_f1)
        print("Best epoch:", best_epoch)

# -------------------- 4. CREATE DATAFRAMES --------------------

summary_df = pd.DataFrame(summary_rows)
epoch_df = pd.DataFrame(epoch_rows)
files_df = pd.DataFrame(file_rows)

# -------------------- 5. SORT RESULTS --------------------

summary_df = summary_df.sort_values(
    ["model", "data_percentage"]
)

if not epoch_df.empty:
    epoch_df = epoch_df.sort_values(
        ["model", "data_percentage", "epoch"],
        na_position="last"
    )

# -------------------- 6. COMPARISON TABLE --------------------

comparison_df = summary_df.pivot(
    index="data_percentage",
    columns="model",
    values="macro_f1"
).reset_index()

# -------------------- 7. BEST EXPERIMENT --------------------

valid_results = summary_df.dropna(
    subset=["macro_f1"]
)

if not valid_results.empty:

    best_row = valid_results.loc[
        valid_results["macro_f1"].idxmax()
    ]

    print("\n" + "=" * 65)
    print("BEST BASELINE EXPERIMENT")
    print("=" * 65)

    print("Model:", best_row["model"])
    print("Training data:", best_row["data_percentage"], "%")
    print("Validation Macro-F1:", best_row["macro_f1"])

# -------------------- 8. 100% PERFORMANCE DROP AUDIT --------------------

print("\n" + "=" * 65)
print("100% DATA PERFORMANCE CHECK")
print("=" * 65)

for model_name in MODELS:

    model_results = summary_df[
        summary_df["model"] == model_name
    ]

    best_partial = model_results[
        model_results["data_percentage"].isin([25, 50, 75])
    ]

    full_data = model_results[
        model_results["data_percentage"] == 100
    ]

    if (
        not best_partial.empty
        and not full_data.empty
        and full_data["macro_f1"].notna().any()
        and best_partial["macro_f1"].notna().any()
    ):

        best_partial_row = best_partial.loc[
            best_partial["macro_f1"].idxmax()
        ]

        full_row = full_data.iloc[0]

        difference = (
            full_row["macro_f1"]
            - best_partial_row["macro_f1"]
        )

        print(
            f"{model_name}: 100% vs best smaller subset = "
            f"{difference * 100:.2f} percentage points"
        )

        if difference < 0:
            print("Observation: Full-data performance is lower.")
        else:
            print("Observation: Full-data performance is higher.")

# -------------------- 9. SAVE REPORTS --------------------

summary_csv = OUTPUT_DIR / "Baseline_Experiment_Audit.csv"
epoch_csv = OUTPUT_DIR / "Baseline_Epoch_Logs.csv"
files_csv = OUTPUT_DIR / "Baseline_Model_File_Audit.csv"

summary_df.to_csv(summary_csv, index=False)
epoch_df.to_csv(epoch_csv, index=False)
files_df.to_csv(files_csv, index=False)

excel_path = OUTPUT_DIR / "TeluguShield_Complete_Baseline_Audit.xlsx"

with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:

    summary_df.to_excel(
        writer,
        sheet_name="Experiment_Summary",
        index=False
    )

    comparison_df.to_excel(
        writer,
        sheet_name="Model_Comparison",
        index=False
    )

    epoch_df.to_excel(
        writer,
        sheet_name="Epoch_Training_Logs",
        index=False
    )

    files_df.to_excel(
        writer,
        sheet_name="Model_File_Audit",
        index=False
    )

# -------------------- 10. FINAL DISPLAY --------------------

print("\n" + "=" * 65)
print("COMPLETE AUDIT SUMMARY")
print("=" * 65)

print(
    summary_df[
        [
            "model",
            "data_percentage",
            "accuracy",
            "macro_f1",
            "best_epoch",
            "last_training_loss",
            "last_validation_loss",
            "trainer_state_found",
            "model_saved",
            "audit_status"
        ]
    ].to_string(index=False)
)

print("\nMODEL COMPARISON - MACRO F1")
print(comparison_df.to_string(index=False))

print("\nEpoch log rows found:", len(epoch_df))

print("\nREPORTS SAVED:")
print(excel_path)
print(summary_csv)
print(epoch_csv)
print(files_csv)

print("\nAUDIT COMPLETED.")
```

```python

# ============================================================
# TELUGUBERT HYPERPARAMETER TUNING - TELUGUSHIELD AI
# 3 learning-rate experiments on the full training pool
# ============================================================

import os, json, shutil, gc
import pandas as pd
import numpy as np
import torch

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    EarlyStoppingCallback
)

# -------------------- 1. PATHS --------------------

BASE = "/content/drive/MyDrive/TeluguShield_AI"

DATA_PATH = os.path.join(
    BASE, "01_Datasets/Processed/telugu_hate_train_cleaned.csv"
)

SAVE_DIR = os.path.join(BASE, "05_Hyperparameter_Tuning/TeluguBERT")
TEMP_DIR = "/content/telugushield_telugubert_tuning"

os.makedirs(SAVE_DIR, exist_ok=True)
os.makedirs(TEMP_DIR, exist_ok=True)

MODEL_NAME = "l3cube-pune/telugu-bert"

# -------------------- 2. CONFIGURATION --------------------

LEARNING_RATES = [1e-5, 2e-5, 3e-5]

EPOCHS = 5
BATCH_SIZE = 8
EVAL_BATCH_SIZE = 16
MAX_LENGTH = 128

SEED = 42
WEIGHT_DECAY = 0.01

print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "No GPU")

# -------------------- 3. LOAD DATA --------------------

df = pd.read_csv(DATA_PATH)

# Automatically identify text and label columns
text_col = next(
    (c for c in df.columns if c.lower() in ["comments", "comment", "text"]),
    None
)
label_col = next(
    (c for c in df.columns if c.lower() in ["label", "labels", "category"]),
    None
)

if text_col is None or label_col is None:
    raise ValueError(f"Text/label columns not found. Available columns: {list(df.columns)}")

df = df[[text_col, label_col]].dropna().copy()
df[text_col] = df[text_col].astype(str).str.strip()

# Consistent binary label mapping
def normalize_label(x):
    x = str(x).strip().lower()
    if x in ["non-hate", "non hate", "non_hate", "not hate", "0"]:
        return 0
    if x in ["hate", "1"]:
        return 1
    raise ValueError(f"Unexpected label found: {x}")

df["labels"] = df[label_col].apply(normalize_label)
df = df[df[text_col] != ""].reset_index(drop=True)

# -------------------- 4. FIXED TRAIN / VALIDATION SPLIT --------------------

train_df, val_df = train_test_split(
    df,
    test_size=0.20,
    random_state=SEED,
    stratify=df["labels"]
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)

print("\nDataset:")
print("Training samples:", len(train_df))
print("Validation samples:", len(val_df))
print("Training distribution:")
print(train_df["labels"].value_counts().sort_index())
print("Validation distribution:")
print(val_df["labels"].value_counts().sort_index())

# -------------------- 5. TOKENIZATION --------------------

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_function(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        padding="max_length",
        max_length=MAX_LENGTH
    )

train_dataset = Dataset.from_dict({
    "text": train_df[text_col].tolist(),
    "labels": train_df["labels"].tolist()
})

val_dataset = Dataset.from_dict({
    "text": val_df[text_col].tolist(),
    "labels": val_df["labels"].tolist()
})

train_dataset = train_dataset.map(tokenize_function, batched=True)
val_dataset = val_dataset.map(tokenize_function, batched=True)

train_dataset = train_dataset.remove_columns(["text"])
val_dataset = val_dataset.remove_columns(["text"])

# -------------------- 6. EVALUATION METRICS --------------------

def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = np.argmax(predictions, axis=1)

    precision, recall, macro_f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="macro",
        zero_division=0
    )

    accuracy = accuracy_score(labels, predictions)

    return {
        "accuracy": accuracy,
        "macro_f1": macro_f1,
        "precision": precision,
        "recall": recall
    }

# -------------------- 7. RUN EXPERIMENTS --------------------

results = []
overall_best_f1 = -1
overall_best_model = os.path.join(SAVE_DIR, "Best_Model")

for trial_number, lr in enumerate(LEARNING_RATES, start=1):

    print("\n" + "="*65)
    print(f"TRIAL {trial_number}/{len(LEARNING_RATES)}")
    print("Learning Rate:", lr)
    print("="*65)

    trial_name = f"TeluguBERT_LR_{lr}"
    trial_output = os.path.join(TEMP_DIR, trial_name)

    if os.path.exists(trial_output):
        shutil.rmtree(trial_output)

    os.makedirs(trial_output, exist_ok=True)

    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=2,
        id2label={0: "non-hate", 1: "hate"},
        label2id={"non-hate": 0, "hate": 1}
    )

    training_args = TrainingArguments(
        output_dir=trial_output,
        learning_rate=lr,
        weight_decay=WEIGHT_DECAY,
        num_train_epochs=EPOCHS,
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=EVAL_BATCH_SIZE,
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="eval_macro_f1",
        greater_is_better=True,
        save_total_limit=1,
        logging_strategy="epoch",
        fp16=torch.cuda.is_available(),
        report_to="none",
        seed=SEED,
        data_seed=SEED
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        processing_class=tokenizer,
        compute_metrics=compute_metrics,
        callbacks=[EarlyStoppingCallback(early_stopping_patience=2)]
    )

    trainer.train()

    # Evaluate the best checkpoint loaded by Trainer
    evaluation = trainer.evaluate()

    # Preserve complete epoch-wise logs before removing checkpoints
    log_history = trainer.state.log_history

    with open(os.path.join(SAVE_DIR, f"{trial_name}_logs.json"), "w") as f:
        json.dump(log_history, f, indent=4, default=str)

    best_f1 = float(evaluation["eval_macro_f1"])

    eval_rows = [
        x for x in log_history
        if "eval_macro_f1" in x
    ]

    best_epoch = (
        max(eval_rows, key=lambda x: x["eval_macro_f1"]).get("epoch")
        if eval_rows else None
    )

    trial_result = {
        "trial": trial_name,
        "learning_rate": lr,
        "weight_decay": WEIGHT_DECAY,
        "epochs_completed": trainer.state.epoch,
        "best_epoch": best_epoch,
        "validation_accuracy": evaluation["eval_accuracy"],
        "validation_macro_f1": best_f1,
        "validation_precision": evaluation["eval_precision"],
        "validation_recall": evaluation["eval_recall"],
        "best_checkpoint": trainer.state.best_model_checkpoint
    }

    results.append(trial_result)

    # Save the overall best model only
    if best_f1 > overall_best_f1:
        overall_best_f1 = best_f1

        if os.path.exists(overall_best_model):
            shutil.rmtree(overall_best_model)

        trainer.save_model(overall_best_model)
        tokenizer.save_pretrained(overall_best_model)

        with open(os.path.join(SAVE_DIR, "best_model_config.json"), "w") as f:
            json.dump(trial_result, f, indent=4, default=str)

        print("NEW BEST MODEL SAVED!")

    print("\nTrial result:")
    print(json.dumps(trial_result, indent=4, default=str))

    # Clean temporary checkpoints and free GPU memory
    del trainer, model
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    shutil.rmtree(trial_output, ignore_errors=True)

    # Save intermediate results after every trial
    pd.DataFrame(results).to_csv(
        os.path.join(SAVE_DIR, "TeluguBERT_Tuning_Results.csv"),
        index=False
    )

# -------------------- 8. FINAL RESULTS --------------------

results_df = pd.DataFrame(results).sort_values(
    "validation_macro_f1",
    ascending=False
)

results_df.to_csv(
    os.path.join(SAVE_DIR, "TeluguBERT_Tuning_Results.csv"),
    index=False
)

results_df.to_excel(
    os.path.join(SAVE_DIR, "TeluguBERT_Tuning_Results.xlsx"),
    index=False
)

print("\n" + "="*65)
print("HYPERPARAMETER TUNING COMPLETED")
print("="*65)

display(results_df)

print("\nBest validation Macro-F1:", round(overall_best_f1, 4))
print("Best model location:", overall_best_model)
print("Results location:", SAVE_DIR)
print("\nBest model exists:", os.path.exists(overall_best_model))
```

```python

# ============================================================
# TELUGUSHIELD AI - MuRIL HYPERPARAMETER TUNING
# 3 learning rates | Full training pool | Fixed validation set
# ============================================================

import os
import gc
import json
import shutil
import random
import numpy as np
import pandas as pd
import torch

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    EarlyStoppingCallback
)

# -------------------- 1. REPRODUCIBILITY --------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# -------------------- 2. PATHS --------------------

BASE = "/content/drive/MyDrive/TeluguShield_AI"

DATA_PATH = os.path.join(
    BASE,
    "01_Datasets/Processed/telugu_hate_train_cleaned.csv"
)

SAVE_DIR = os.path.join(
    BASE,
    "05_Hyperparameter_Tuning/MuRIL"
)

TEMP_DIR = "/content/telugushield_muril_tuning"

os.makedirs(SAVE_DIR, exist_ok=True)
os.makedirs(TEMP_DIR, exist_ok=True)

MODEL_NAME = "google/muril-base-cased"

# -------------------- 3. EXPERIMENT SETTINGS --------------------

LEARNING_RATES = [1e-5, 2e-5, 3e-5]

EPOCHS = 5
BATCH_SIZE = 8
EVAL_BATCH_SIZE = 16
MAX_LENGTH = 128
WEIGHT_DECAY = 0.01

print("=" * 60)
print("MuRIL HYPERPARAMETER TUNING")
print("=" * 60)

print(
    "GPU:",
    torch.cuda.get_device_name(0)
    if torch.cuda.is_available()
    else "No GPU"
)

# -------------------- 4. LOAD CLEANED DATA --------------------

if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(f"Dataset not found: {DATA_PATH}")

df = pd.read_csv(DATA_PATH)

text_col = next(
    (c for c in df.columns
     if c.lower() in ["comments", "comment", "text"]),
    None
)

label_col = next(
    (c for c in df.columns
     if c.lower() in ["label", "labels", "category"]),
    None
)

if text_col is None or label_col is None:
    raise ValueError(
        f"Required columns not found. Available: {list(df.columns)}"
    )

df = df[[text_col, label_col]].dropna().copy()

df[text_col] = df[text_col].astype(str).str.strip()

def normalize_label(value):
    value = str(value).strip().lower()

    if value in ["non-hate", "non hate", "non_hate",
                 "not hate", "0"]:
        return 0

    if value in ["hate", "1"]:
        return 1

    raise ValueError(f"Unexpected label: {value}")

df["labels"] = df[label_col].apply(normalize_label)

df = df[df[text_col] != ""].reset_index(drop=True)

# -------------------- 5. FIXED TRAIN / VALIDATION SPLIT --------------------

train_df, val_df = train_test_split(
    df,
    test_size=0.20,
    random_state=SEED,
    stratify=df["labels"]
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)

print("\nDataset information:")
print("Total samples:", len(df))
print("Training samples:", len(train_df))
print("Validation samples:", len(val_df))

print("\nTraining label distribution:")
print(train_df["labels"].value_counts().sort_index())

print("\nValidation label distribution:")
print(val_df["labels"].value_counts().sort_index())

# -------------------- 6. TOKENIZATION --------------------

print("\nLoading MuRIL tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def tokenize_function(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        padding="max_length",
        max_length=MAX_LENGTH
    )

train_dataset = Dataset.from_dict({
    "text": train_df[text_col].tolist(),
    "labels": train_df["labels"].tolist()
})

val_dataset = Dataset.from_dict({
    "text": val_df[text_col].tolist(),
    "labels": val_df["labels"].tolist()
})

train_dataset = train_dataset.map(
    tokenize_function,
    batched=True
)

val_dataset = val_dataset.map(
    tokenize_function,
    batched=True
)

train_dataset = train_dataset.remove_columns(["text"])
val_dataset = val_dataset.remove_columns(["text"])

# -------------------- 7. EVALUATION METRICS --------------------

def compute_metrics(eval_pred):

    predictions, labels = eval_pred

    predictions = np.argmax(predictions, axis=1)

    precision, recall, macro_f1, _ = (
        precision_recall_fscore_support(
            labels,
            predictions,
            average="macro",
            zero_division=0
        )
    )

    accuracy = accuracy_score(labels, predictions)

    return {
        "accuracy": accuracy,
        "macro_f1": macro_f1,
        "precision": precision,
        "recall": recall
    }

# -------------------- 8. RUN THREE EXPERIMENTS --------------------

results = []

overall_best_f1 = -1.0

BEST_MODEL_DIR = os.path.join(SAVE_DIR, "Best_Model")

for trial_number, lr in enumerate(LEARNING_RATES, start=1):

    print("\n" + "=" * 65)
    print(f"TRIAL {trial_number}/{len(LEARNING_RATES)}")
    print("Learning rate:", lr)
    print("=" * 65)

    trial_name = f"MuRIL_LR_{lr}"

    trial_output = os.path.join(TEMP_DIR, trial_name)

    if os.path.exists(trial_output):
        shutil.rmtree(trial_output)

    os.makedirs(trial_output, exist_ok=True)

    # Start every trial from the original pretrained MuRIL
    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=2,
        id2label={
            0: "non-hate",
            1: "hate"
        },
        label2id={
            "non-hate": 0,
            "hate": 1
        }
    )

    training_args = TrainingArguments(
        output_dir=trial_output,

        learning_rate=lr,
        weight_decay=WEIGHT_DECAY,

        num_train_epochs=EPOCHS,

        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=EVAL_BATCH_SIZE,

        eval_strategy="epoch",
        save_strategy="epoch",

        load_best_model_at_end=True,
        metric_for_best_model="eval_macro_f1",
        greater_is_better=True,

        save_total_limit=1,

        logging_strategy="epoch",

        fp16=torch.cuda.is_available(),

        report_to="none",

        seed=SEED,
        data_seed=SEED
    )

    trainer = Trainer(
        model=model,
        args=training_args,

        train_dataset=train_dataset,
        eval_dataset=val_dataset,

        processing_class=tokenizer,

        compute_metrics=compute_metrics,

        callbacks=[
            EarlyStoppingCallback(
                early_stopping_patience=2
            )
        ]
    )

    # Train model
    trainer.train()

    # Evaluate the best checkpoint
    evaluation = trainer.evaluate()

    # Preserve epoch-wise logs
    log_history = trainer.state.log_history

    log_path = os.path.join(
        SAVE_DIR,
        f"{trial_name}_logs.json"
    )

    with open(log_path, "w") as f:
        json.dump(
            log_history,
            f,
            indent=4,
            default=str
        )

    # Find best epoch from evaluation logs
    eval_rows = [
        row for row in log_history
        if "eval_macro_f1" in row
    ]

    best_epoch = (
        max(
            eval_rows,
            key=lambda row: row["eval_macro_f1"]
        ).get("epoch")
        if eval_rows else None
    )

    best_f1 = float(evaluation["eval_macro_f1"])

    trial_result = {
        "trial": trial_name,
        "learning_rate": lr,
        "weight_decay": WEIGHT_DECAY,
        "epochs_completed": trainer.state.epoch,
        "best_epoch": best_epoch,
        "validation_accuracy": float(evaluation["eval_accuracy"]),
        "validation_macro_f1": best_f1,
        "validation_precision": float(evaluation["eval_precision"]),
        "validation_recall": float(evaluation["eval_recall"])
    }

    results.append(trial_result)

    # Save best model from all trials
    if best_f1 > overall_best_f1:

        overall_best_f1 = best_f1

        if os.path.exists(BEST_MODEL_DIR):
            shutil.rmtree(BEST_MODEL_DIR)

        trainer.save_model(BEST_MODEL_DIR)
        tokenizer.save_pretrained(BEST_MODEL_DIR)

        with open(
            os.path.join(SAVE_DIR, "best_model_config.json"),
            "w"
        ) as f:
            json.dump(
                trial_result,
                f,
                indent=4,
                default=str
            )

        print("\nNEW BEST MuRIL MODEL SAVED!")

    print("\nTrial result:")
    print(json.dumps(trial_result, indent=4))

    # Save intermediate results after every trial
    results_df = pd.DataFrame(results)

    results_df.to_csv(
        os.path.join(
            SAVE_DIR,
            "MuRIL_Tuning_Results.csv"
        ),
        index=False
    )

    # Free memory and remove temporary checkpoints
    del trainer
    del model

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    shutil.rmtree(trial_output, ignore_errors=True)

# -------------------- 9. FINAL RESULTS --------------------

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    "validation_macro_f1",
    ascending=False
).reset_index(drop=True)

results_df.to_csv(
    os.path.join(
        SAVE_DIR,
        "MuRIL_Tuning_Results.csv"
    ),
    index=False
)

results_df.to_excel(
    os.path.join(
        SAVE_DIR,
        "MuRIL_Tuning_Results.xlsx"
    ),
    index=False
)

print("\n" + "=" * 65)
print("MURIL HYPERPARAMETER TUNING COMPLETED")
print("=" * 65)

display(results_df)

print(
    "\nBest validation Macro-F1:",
    round(overall_best_f1, 4)
)

print("\nBest model path:")
print(BEST_MODEL_DIR)

print("\nResults folder:")
print(SAVE_DIR)

print(
    "\nBest model exists:",
    os.path.exists(BEST_MODEL_DIR)
)
```

```python

import pandas as pd
import os

BASE = "/content/drive/MyDrive/TeluguShield_AI"

test_path = f"{BASE}/01_Datasets/Processed/telugu_hate_test_cleaned.xlsx"
review_path = f"{BASE}/02_Preprocessing/Manual_Label_Review.xlsx"

output_xlsx = f"{BASE}/01_Datasets/Processed/telugu_hate_test_final.xlsx"
output_csv = f"{BASE}/01_Datasets/Processed/telugu_hate_test_final.csv"

# Load test dataset and approved manual review
test_df = pd.read_excel(test_path)
review_df = pd.read_excel(
    review_path,
    sheet_name="Cross_Dataset_Conflict"
)

# Preserve original Label; create separate Final_Label
test_df["Final_Label"] = test_df["Label"].astype(str).str.strip().str.lower()

# Apply approved cross-dataset decision
def normalize(text):
    return " ".join(str(text).lower().split())

for _, row in review_df.iterrows():
    comment = normalize(row["Comment"])
    decision = str(row["Final_Decision"]).strip().lower()

    mask = test_df["Comments"].apply(normalize) == comment

    if mask.any():
        test_df.loc[mask, "Final_Label"] = decision
        print("Applied approved decision:", comment, "->", decision)

# Validate labels
valid_labels = {"hate", "non-hate"}

assert set(test_df["Final_Label"].unique()).issubset(valid_labels)
assert test_df["Final_Label"].notna().all()
assert len(test_df) == 489

# Save new files; original cleaned files remain untouched
test_df.to_excel(output_xlsx, index=False)
test_df.to_csv(output_csv, index=False)

print("\nFINAL TEST DATASET")
print("Shape:", test_df.shape)
print("Columns:", test_df.columns.tolist())
print("\nFinal label distribution:")
print(test_df["Final_Label"].value_counts())

print("\nOriginal Label distribution:")
print(test_df["Label"].value_counts())

print("\nMissing values:", test_df.isnull().sum().sum())
print("\nExcel saved:", os.path.exists(output_xlsx))
print("CSV saved:", os.path.exists(output_csv))
```

```python

import os
import gc
import time
import json
import numpy as np
import pandas as pd
import torch

from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ==========================================
# 1. PATHS AND CONFIGURATION
# ==========================================

BASE = "/content/drive/MyDrive/TeluguShield_AI"

TEST_PATH = f"{BASE}/01_Datasets/Processed/telugu_hate_test_final.csv"

RESULT_DIR = f"{BASE}/06_Final_Evaluation"
os.makedirs(RESULT_DIR, exist_ok=True)

MODEL_PATHS = {
    "MuRIL": f"{BASE}/05_Hyperparameter_Tuning/MuRIL/Best_Model",
    "TeluguBERT": f"{BASE}/05_Hyperparameter_Tuning/TeluguBERT/Best_Model",
    "XLM-RoBERTa": f"{BASE}/05_Hyperparameter_Tuning/XLM-RoBERTa/Best_Model"
}

MAX_LENGTH = 128
BATCH_SIZE = 16

LABEL_MAP = {
    "non-hate": 0,
    "hate": 1
}

ID_TO_LABEL = {
    0: "non-hate",
    1: "hate"
}

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

# ==========================================
# 2. LOAD INDEPENDENT TEST DATASET
# ==========================================

test_df = pd.read_csv(TEST_PATH)

assert len(test_df) == 489
assert "Comments" in test_df.columns
assert "Final_Label" in test_df.columns
assert test_df["Comments"].notna().all()
assert test_df["Final_Label"].notna().all()

X_test = test_df["Comments"].astype(str).tolist()

y_test = (
    test_df["Final_Label"]
    .str.strip()
    .str.lower()
    .map(LABEL_MAP)
    .to_numpy()
)

assert not np.isnan(y_test).any()

print("\nTest dataset:", len(X_test))
print("Label distribution:")
print(test_df["Final_Label"].value_counts())

# ==========================================
# 3. EVALUATE EACH MODEL
# ==========================================

all_results = []
all_predictions = {}
all_reports = {}

for model_name, model_path in MODEL_PATHS.items():

    print("\n" + "=" * 60)
    print("EVALUATING:", model_name)
    print("=" * 60)

    assert os.path.isdir(model_path), f"Missing model: {model_path}"

    tokenizer = AutoTokenizer.from_pretrained(model_path)

    model = AutoModelForSequenceClassification.from_pretrained(
        model_path
    )

    model.to(device)
    model.eval()

    print("Model loaded successfully.")

    predictions = []
    probabilities = []
    total_inference_time = 0

    # Process in batches
    for start in range(0, len(X_test), BATCH_SIZE):

        batch_texts = X_test[start:start + BATCH_SIZE]

        inputs = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=MAX_LENGTH,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
        }

        if device.type == "cuda":
            torch.cuda.synchronize()

        start_time = time.perf_counter()

        with torch.inference_mode():
            outputs = model(**inputs)
            probs = torch.softmax(outputs.logits, dim=-1)
            batch_predictions = torch.argmax(probs, dim=-1)

        if device.type == "cuda":
            torch.cuda.synchronize()

        total_inference_time += time.perf_counter() - start_time

        predictions.extend(batch_predictions.cpu().numpy())
        probabilities.extend(probs.cpu().numpy())

    predictions = np.array(predictions)
    probabilities = np.array(probabilities)

    # ======================================
    # 4. CALCULATE METRICS
    # ======================================

    accuracy = accuracy_score(y_test, predictions)

    precision = precision_score(
        y_test, predictions, average="macro", zero_division=0
    )

    recall = recall_score(
        y_test, predictions, average="macro", zero_division=0
    )

    macro_f1 = f1_score(
        y_test, predictions, average="macro", zero_division=0
    )

    weighted_f1 = f1_score(
        y_test, predictions, average="weighted", zero_division=0
    )

    cm = confusion_matrix(y_test, predictions, labels=[0, 1])

    avg_inference_ms = (
        total_inference_time / len(X_test)
    ) * 1000

    result = {
        "Model": model_name,
        "Accuracy": accuracy,
        "Macro_Precision": precision,
        "Macro_Recall": recall,
        "Macro_F1": macro_f1,
        "Weighted_F1": weighted_f1,
        "Total_Inference_Time_Seconds": total_inference_time,
        "Average_Inference_Time_ms": avg_inference_ms
    }

    all_results.append(result)
    all_predictions[model_name] = predictions

    # Classification report
    report = classification_report(
        y_test,
        predictions,
        labels=[0, 1],
        target_names=["non-hate", "hate"],
        output_dict=True,
        zero_division=0
    )

    all_reports[model_name] = report

    # Save confusion matrix
    cm_df = pd.DataFrame(
        cm,
        index=["Actual_non-hate", "Actual_hate"],
        columns=["Predicted_non-hate", "Predicted_hate"]
    )

    cm_df.to_csv(
        f"{RESULT_DIR}/{model_name}_confusion_matrix.csv"
    )

    # Save individual predictions
    prediction_df = test_df.copy()

    prediction_df["Predicted_Label"] = [
        ID_TO_LABEL[int(x)] for x in predictions
    ]

    prediction_df["Prediction_Confidence"] = probabilities.max(axis=1)

    prediction_df["Correct_Prediction"] = (
        predictions == y_test
    )

    prediction_df.to_csv(
        f"{RESULT_DIR}/{model_name}_predictions.csv",
        index=False
    )

    print("\nAccuracy:", round(accuracy * 100, 2), "%")
    print("Macro Precision:", round(precision * 100, 2), "%")
    print("Macro Recall:", round(recall * 100, 2), "%")
    print("Macro F1:", round(macro_f1 * 100, 2), "%")
    print("Weighted F1:", round(weighted_f1 * 100, 2), "%")

    print("\nConfusion Matrix:")
    print(cm)

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        predictions,
        labels=[0, 1],
        target_names=["non-hate", "hate"],
        zero_division=0
    ))

    # Release memory before next model
    del model, tokenizer
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# ==========================================
# 5. SAVE COMPARATIVE RESULTS
# ==========================================

results_df = pd.DataFrame(all_results)

results_df = results_df.sort_values(
    by="Macro_F1",
    ascending=False
).reset_index(drop=True)

results_df.insert(
    0,
    "Rank",
    range(1, len(results_df) + 1)
)

results_df.to_csv(
    f"{RESULT_DIR}/Final_Model_Comparison.csv",
    index=False
)

results_df.to_excel(
    f"{RESULT_DIR}/Final_Model_Comparison.xlsx",
    index=False
)

# Save detailed classification reports
with open(
    f"{RESULT_DIR}/Final_Classification_Reports.json",
    "w"
) as f:
    json.dump(all_reports, f, indent=4)

# ==========================================
# 6. FINAL SUMMARY
# ==========================================

print("\n" + "=" * 70)
print("TELUGUSHIELD AI - FINAL INDEPENDENT TEST RESULTS")
print("=" * 70)

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Macro_Precision": "{:.4f}",
        "Macro_Recall": "{:.4f}",
        "Macro_F1": "{:.4f}",
        "Weighted_F1": "{:.4f}",
        "Average_Inference_Time_ms": "{:.3f}"
    })
)

print("\nBest model by test Macro-F1:",
      results_df.iloc[0]["Model"])

print("\nAll evaluation files saved in:")
print(RESULT_DIR)

print("\nIndependent test evaluation completed.")
```

```python
!pip -q install transformers scikit-learn openpyxl

import os
import time
import json
import torch
import pandas as pd
import numpy as np

from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
from torch.utils.data import Dataset, DataLoader

# Paths
BASE = "/content/gdrive/MyDrive/TeluguShield_AI"

TEST_PATH = f"{BASE}/01_Datasets/Processed/telugu_hate_test_final.csv"
OUTPUT_DIR = f"{BASE}/06_Final_Evaluation"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

# Load test dataset
test_df = pd.read_csv(TEST_PATH)

texts = test_df["Comments"].astype(str).tolist()

label_map = {
    "non-hate": 0,
    "hate": 1
}

y_true = test_df["Final_Label"].str.lower().map(label_map).values

assert len(texts) == 489
assert not pd.isna(y_true).any()

# Model paths
models = {
    "MuRIL": f"{BASE}/05_Hyperparameter_Tuning/MuRIL/Best_Model",
    "TeluguBERT": f"{BASE}/05_Hyperparameter_Tuning/TeluguBERT/Best_Model",
    "XLM-RoBERTa": f"{BASE}/05_Hyperparameter_Tuning/XLM-RoBERTa/Best_Model"
}

MAX_LENGTH = 128
BATCH_SIZE = 16

all_results = []
all_predictions = {}

for model_name, model_path in models.items():

    print("\n" + "=" * 60)
    print("Evaluating:", model_name)
    print("=" * 60)

    tokenizer = AutoTokenizer.from_pretrained(model_path)

    model = AutoModelForSequenceClassification.from_pretrained(
        model_path
    ).to(device)

    model.eval()

    predictions = []
    probabilities = []

    inference_start = time.time()

    for start in range(0, len(texts), BATCH_SIZE):

        batch_texts = texts[start:start+BATCH_SIZE]

        inputs = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=MAX_LENGTH,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
        }

        with torch.no_grad():
            outputs = model(**inputs)
            probs = torch.softmax(outputs.logits, dim=1)

        predictions.extend(
            torch.argmax(probs, dim=1).cpu().numpy()
        )

        probabilities.extend(
            probs.cpu().numpy()
        )

    if torch.cuda.is_available():
        torch.cuda.synchronize()

    inference_time = time.time() - inference_start

    predictions = np.array(predictions)
    probabilities = np.array(probabilities)

    # Metrics
    accuracy = accuracy_score(y_true, predictions)
    precision = precision_score(
        y_true, predictions, average="macro", zero_division=0
    )
    recall = recall_score(
        y_true, predictions, average="macro", zero_division=0
    )
    macro_f1 = f1_score(
        y_true, predictions, average="macro", zero_division=0
    )
    weighted_f1 = f1_score(
        y_true, predictions, average="weighted", zero_division=0
    )

    result = {
        "Model": model_name,
        "Accuracy": accuracy,
        "Macro_Precision": precision,
        "Macro_Recall": recall,
        "Macro_F1": macro_f1,
        "Weighted_F1": weighted_f1,
        "Inference_Time_Seconds": inference_time,
        "Average_Inference_ms_per_Comment":
            (inference_time / len(texts)) * 1000
    }

    all_results.append(result)

    all_predictions[model_name] = predictions

    # Detailed classification report
    report = classification_report(
        y_true,
        predictions,
        labels=[0, 1],
        target_names=["non-hate", "hate"],
        output_dict=True,
        zero_division=0
    )

    pd.DataFrame(report).transpose().to_csv(
        f"{OUTPUT_DIR}/{model_name}_Classification_Report.csv"
    )

    # Confusion matrix
    cm = confusion_matrix(y_true, predictions, labels=[0, 1])

    pd.DataFrame(
        cm,
        index=["Actual_non-hate", "Actual_hate"],
        columns=["Predicted_non-hate", "Predicted_hate"]
    ).to_csv(
        f"{OUTPUT_DIR}/{model_name}_Confusion_Matrix.csv"
    )

    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"Macro-F1:  {macro_f1:.4f}")
    print(f"Weighted F1: {weighted_f1:.4f}")
    print(f"Inference time: {inference_time:.2f} seconds")

    del model, tokenizer
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# Comparison
results_df = pd.DataFrame(all_results)

results_df.to_csv(
    f"{OUTPUT_DIR}/Final_Model_Comparison.csv",
    index=False
)

results_df.to_excel(
    f"{OUTPUT_DIR}/Final_Model_Comparison.xlsx",
    index=False
)

# Comment-level predictions
prediction_df = test_df.copy()

for model_name, preds in all_predictions.items():

    prediction_df[f"{model_name}_Prediction"] = [
        "hate" if p == 1 else "non-hate"
        for p in preds
    ]

prediction_df.to_csv(
    f"{OUTPUT_DIR}/All_Model_Test_Predictions.csv",
    index=False
)

print("\n" + "=" * 60)
print("FINAL MODEL COMPARISON")
print("=" * 60)

print(results_df.to_string(index=False))

print("\nAll evaluation files saved in:")
print(OUTPUT_DIR)
```

```python

# ================================================================
# TELUGUSHIELD AI — COMPLETE XAI IMPLEMENTATION
# Explainability Method: Integrated Gradients
# Models: MuRIL, TeluguBERT, XLM-RoBERTa
# Dataset: Independent Final Test Dataset
# ================================================================

!pip -q install captum openpyxl

import os
import gc
import html
import torch
import numpy as np
import pandas as pd

from captum.attr import LayerIntegratedGradients
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from IPython.display import display, HTML

# ================================================================
# 1. CONFIGURATION
# ================================================================

BASE = "/content/gdrive/MyDrive/TeluguShield_AI"

TEST_FILE = (
    f"{BASE}/01_Datasets/Processed/"
    "telugu_hate_test_final.csv"
)

XAI_DIR = f"{BASE}/07_Explainability"

MODEL_PATHS = {
    "MuRIL": f"{BASE}/05_Hyperparameter_Tuning/MuRIL/Best_Model",
    "TeluguBERT": f"{BASE}/05_Hyperparameter_Tuning/TeluguBERT/Best_Model",
    "XLM-RoBERTa": f"{BASE}/05_Hyperparameter_Tuning/XLM-RoBERTa/Best_Model"
}

os.makedirs(XAI_DIR, exist_ok=True)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

MAX_LENGTH = 128
BATCH_SIZE = 16
IG_STEPS = 32

print("=" * 70)
print("TELUGUSHIELD AI — INTEGRATED GRADIENTS")
print("=" * 70)
print("Device:", DEVICE)

if DEVICE.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# ================================================================
# 2. LOAD AND VALIDATE TEST DATA
# ================================================================

if not os.path.exists(TEST_FILE):
    raise FileNotFoundError(f"Test dataset not found: {TEST_FILE}")

df = pd.read_csv(TEST_FILE)

required_columns = ["Comments", "Final_Label"]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

df = df.dropna(subset=required_columns).reset_index(drop=True)

def normalize_label(value):

    value = str(value).strip().lower()

    if value in ["1", "1.0", "hate"]:
        return "hate"

    if value in ["0", "0.0", "non-hate", "non hate", "non_hate"]:
        return "non-hate"

    raise ValueError(f"Unknown label: {value}")

df["True_Label"] = df["Final_Label"].apply(normalize_label)

df["Comments"] = df["Comments"].astype(str)

print("\nTest dataset:", df.shape)
print("\nLabel distribution:")
print(df["True_Label"].value_counts())

# ================================================================
# 3. LABEL MAPPING
# ================================================================

def get_label_mapping(model):

    id2label = model.config.id2label

    mapping = {}

    for class_id in range(model.config.num_labels):

        label = str(
            id2label.get(
                class_id,
                id2label.get(str(class_id), "")
            )
        ).lower().replace("_", "-")

        if "non" in label:
            mapping[class_id] = "non-hate"

        elif "hate" in label:
            mapping[class_id] = "hate"

        else:
            # Fallback based on the original binary label encoding.
            mapping[class_id] = (
                "non-hate" if class_id == 0 else "hate"
            )

    if set(mapping.values()) != {"non-hate", "hate"}:
        raise ValueError(
            f"Could not establish binary label mapping: {mapping}"
        )

    return mapping


# ================================================================
# 4. MODEL PREDICTION FUNCTION
# ================================================================

def get_model_predictions(model, tokenizer, comments, label_mapping):

    predictions = []

    model.eval()

    for start in range(0, len(comments), BATCH_SIZE):

        batch = comments[start:start + BATCH_SIZE]

        inputs = tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=MAX_LENGTH,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(DEVICE)
            for key, value in inputs.items()
        }

        with torch.no_grad():

            outputs = model(**inputs)

            probabilities = torch.softmax(
                outputs.logits,
                dim=-1
            )

        predicted_ids = probabilities.argmax(dim=-1)

        for i in range(len(batch)):

            class_id = int(predicted_ids[i].item())

            predictions.append({
                "Predicted_Class_ID": class_id,
                "Predicted_Label": label_mapping[class_id],
                "Confidence": float(
                    probabilities[i, class_id].item()
                )
            })

    return predictions


# ================================================================
# 5. INTEGRATED GRADIENTS FORWARD FUNCTION
# ================================================================

def make_forward_function(model, has_token_type_ids):

    if has_token_type_ids:

        def forward_func(input_ids, attention_mask, token_type_ids):

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                token_type_ids=token_type_ids
            )

            return outputs.logits

    else:

        def forward_func(input_ids, attention_mask):

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            return outputs.logits

    return forward_func


# ================================================================
# 6. CALCULATE INTEGRATED GRADIENTS
# ================================================================

def calculate_integrated_gradients(
    model,
    tokenizer,
    comment,
    target_class
):

    encoded = tokenizer(
        comment,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_LENGTH,
        padding=False
    )

    input_ids = encoded["input_ids"].to(DEVICE)

    attention_mask = encoded["attention_mask"].to(DEVICE)

    token_type_ids = encoded.get("token_type_ids")

    if token_type_ids is not None:
        token_type_ids = token_type_ids.to(DEVICE)

    # Create baseline by replacing ordinary tokens with PAD.
    baseline_ids = input_ids.clone()

    if tokenizer.pad_token_id is None:
        raise ValueError("Tokenizer has no PAD token.")

    special_ids = set(tokenizer.all_special_ids)

    for position in range(input_ids.shape[1]):

        token_id = input_ids[0, position].item()

        if (
            attention_mask[0, position].item() == 1
            and token_id not in special_ids
        ):

            baseline_ids[0, position] = tokenizer.pad_token_id

    has_token_type_ids = token_type_ids is not None

    forward_func = make_forward_function(
        model,
        has_token_type_ids
    )

    embedding_layer = model.get_input_embeddings()

    lig = LayerIntegratedGradients(
        forward_func,
        embedding_layer
    )

    if has_token_type_ids:

        additional_args = (
            attention_mask,
            token_type_ids
        )

    else:

        additional_args = (
            attention_mask,
        )

    attributions, convergence_delta = lig.attribute(
        inputs=input_ids,
        baselines=baseline_ids,
        additional_forward_args=additional_args,
        target=target_class,
        n_steps=IG_STEPS,
        internal_batch_size=8,
        return_convergence_delta=True
    )

    # Aggregate embedding-dimension attribution into token scores.
    token_scores = (
        attributions.sum(dim=-1)
        .squeeze(0)
        .detach()
        .cpu()
        .numpy()
    )

    token_ids = input_ids[0].detach().cpu().tolist()

    tokens = tokenizer.convert_ids_to_tokens(token_ids)

    attention = attention_mask[0].detach().cpu().numpy()

    records = []

    for token, score, mask in zip(tokens, token_scores, attention):

        if mask == 0:
            continue

        if token in tokenizer.all_special_tokens:
            continue

        records.append({
            "Token": token,
            "Attribution": float(score),
            "Absolute_Attribution": float(abs(score))
        })

    delta = float(
        convergence_delta.detach().cpu().mean().item()
    )

    return records, delta


# ================================================================
# 7. CREATE HTML EXPLANATION
# ================================================================

def create_highlight_html(
    comment,
    records,
    model_name,
    true_label,
    predicted_label,
    confidence,
    category
):

    if not records:
        return "<html><body>No token attribution available.</body></html>"

    max_abs = max(
        [abs(x["Attribution"]) for x in records] + [1e-9]
    )

    token_html = []

    for record in records:

        token = html.escape(record["Token"])

        score = record["Attribution"]

        opacity = min(abs(score) / max_abs, 1.0)

        if score >= 0:
            background = (
                f"rgba(220,70,70,{0.15 + 0.65 * opacity})"
            )
        else:
            background = (
                f"rgba(60,120,220,{0.15 + 0.65 * opacity})"
            )

        token_html.append(
            f'<span title="Attribution: {score:.6f}" '
            f'style="background:{background}; '
            f'padding:5px; margin:3px; '
            f'border-radius:5px; display:inline-block;">'
            f'{token}</span>'
        )

    return f"""
    <!DOCTYPE html>
    <html>
    <head>
    <meta charset="UTF-8">
    <title>{model_name} Integrated Gradients</title>
    </head>

    <body style="font-family:Arial;padding:25px;line-height:2;">

    <h2>{html.escape(model_name)} — Integrated Gradients</h2>

    <h3>Classification details</h3>

    <p><b>Example type:</b> {category}</p>
    <p><b>Original comment:</b> {html.escape(comment)}</p>
    <p><b>Actual label:</b> {true_label}</p>
    <p><b>Predicted label:</b> {predicted_label}</p>
    <p><b>Confidence:</b> {confidence:.4f}</p>

    <h3>Token-level explanation</h3>

    <div>{''.join(token_html)}</div>

    <hr>

    <p>
    <span style="background:rgba(220,70,70,0.7);padding:5px;">
    Positive attribution
    </span>
    Supports the selected predicted class.
    </p>

    <p>
    <span style="background:rgba(60,120,220,0.7);padding:5px;">
    Negative attribution
    </span>
    Opposes the selected predicted class.
    </p>

    <p><i>
    Integrated Gradients measures attribution relative to the selected
    baseline. It does not establish the true linguistic meaning or intent.
    </i></p>

    </body>
    </html>
    """


# ================================================================
# 8. SELECT REPRESENTATIVE EXAMPLES
# ================================================================

def select_examples(model_df):

    conditions = {

        "True_Negative": (
            (model_df["True_Label"] == "non-hate") &
            (model_df["Predicted_Label"] == "non-hate")
        ),

        "False_Positive": (
            (model_df["True_Label"] == "non-hate") &
            (model_df["Predicted_Label"] == "hate")
        ),

        "True_Positive": (
            (model_df["True_Label"] == "hate") &
            (model_df["Predicted_Label"] == "hate")
        ),

        "False_Negative": (
            (model_df["True_Label"] == "hate") &
            (model_df["Predicted_Label"] == "non-hate")
        )
    }

    selected = []

    for category, condition in conditions.items():

        candidates = model_df[condition].copy()

        if candidates.empty:

            print("No example found:", category)
            continue

        # Choose the highest-confidence representative example.
        example = candidates.sort_values(
            "Confidence",
            ascending=False
        ).iloc[0]

        selected.append((category, example))

    return selected


# ================================================================
# 9. PROCESS ALL THREE MODELS
# ================================================================

all_attributions = []
all_predictions = []
all_summaries = []

for model_name, model_path in MODEL_PATHS.items():

    print("\n" + "=" * 70)
    print("PROCESSING MODEL:", model_name)
    print("=" * 70)

    if not os.path.exists(model_path):
        raise FileNotFoundError(
            f"Model directory not found: {model_path}"
        )

    tokenizer = AutoTokenizer.from_pretrained(model_path)

    model = AutoModelForSequenceClassification.from_pretrained(
        model_path
    ).to(DEVICE)

    model.eval()

    label_mapping = get_label_mapping(model)

    print("Label mapping:", label_mapping)

    # Generate predictions for all 489 test comments.
    model_predictions = get_model_predictions(
        model,
        tokenizer,
        df["Comments"].tolist(),
        label_mapping
    )

    # IMPORTANT: Create a fresh DataFrame for this model.
    model_df = df.copy()

    model_df["Predicted_Class_ID"] = [
        item["Predicted_Class_ID"]
        for item in model_predictions
    ]

    model_df["Predicted_Label"] = [
        item["Predicted_Label"]
        for item in model_predictions
    ]

    model_df["Confidence"] = [
        item["Confidence"]
        for item in model_predictions
    ]

    model_df["Correct"] = (
        model_df["True_Label"] ==
        model_df["Predicted_Label"]
    )

    model_df["Model"] = model_name

    all_predictions.append(model_df.copy())

    # Save predictions for each model.
    model_filename = model_name.replace("-", "_")

    model_prediction_path = (
        f"{XAI_DIR}/{model_filename}_XAI_Predictions.csv"
    )

    model_df.to_csv(
        model_prediction_path,
        index=False,
        encoding="utf-8-sig"
    )

    print("Predictions generated:", len(model_df))
    print("Correct predictions:", int(model_df["Correct"].sum()))
    print("Incorrect predictions:", int((~model_df["Correct"]).sum()))

    # Select one example from each confusion-matrix category.
    selected_examples = select_examples(model_df)

    model_output_dir = f"{XAI_DIR}/{model_filename}"

    os.makedirs(model_output_dir, exist_ok=True)

    # ============================================================
    # 10. RUN INTEGRATED GRADIENTS
    # ============================================================

    for category, selected in selected_examples:

        comment = str(selected["Comments"])

        true_label = selected["True_Label"]

        predicted_label = selected["Predicted_Label"]

        confidence = float(selected["Confidence"])

        target_class = int(selected["Predicted_Class_ID"])

        print("\n--------------------------------------------")
        print("Example:", category)
        print("Comment:", comment)
        print("Actual:", true_label)
        print("Predicted:", predicted_label)
        print("Target class ID:", target_class)

        records, convergence_delta = calculate_integrated_gradients(
            model=model,
            tokenizer=tokenizer,
            comment=comment,
            target_class=target_class
        )

        for record in records:

            all_attributions.append({

                "Model": model_name,
                "Example_Type": category,
                "Comment": comment,
                "True_Label": true_label,
                "Predicted_Label": predicted_label,
                "Confidence": confidence,
                "Token": record["Token"],
                "Attribution": record["Attribution"],
                "Absolute_Attribution": record["Absolute_Attribution"],
                "Convergence_Delta": convergence_delta

            })

        # Save individual HTML explanation.
        html_report = create_highlight_html(
            comment,
            records,
            model_name,
            true_label,
            predicted_label,
            confidence,
            category
        )

        html_path = (
            f"{model_output_dir}/"
            f"{category.lower()}_integrated_gradients.html"
        )

        with open(html_path, "w", encoding="utf-8") as file:
            file.write(html_report)

        all_summaries.append({

            "Model": model_name,
            "Example_Type": category,
            "Comment": comment,
            "True_Label": true_label,
            "Predicted_Label": predicted_label,
            "Confidence": confidence,
            "Target_Class_ID": target_class,
            "Number_of_Tokens": len(records),
            "Convergence_Delta": convergence_delta,
            "HTML_Report": html_path

        })

        print("Tokens analysed:", len(records))
        print("Convergence delta:", convergence_delta)
        print("HTML saved:", html_path)

    # Release memory before loading the next model.
    del model
    del tokenizer

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

# ================================================================
# 11. COMBINE ALL RESULTS
# ================================================================

attribution_df = pd.DataFrame(all_attributions)

summary_df = pd.DataFrame(all_summaries)

predictions_df = pd.concat(
    all_predictions,
    ignore_index=True
)

# Save CSV files.
attribution_csv = (
    f"{XAI_DIR}/Integrated_Gradients_All_Models.csv"
)

summary_csv = (
    f"{XAI_DIR}/XAI_Example_Summary.csv"
)

predictions_csv = (
    f"{XAI_DIR}/XAI_All_Model_Predictions.csv"
)

attribution_df.to_csv(
    attribution_csv,
    index=False,
    encoding="utf-8-sig"
)

summary_df.to_csv(
    summary_csv,
    index=False,
    encoding="utf-8-sig"
)

predictions_df.to_csv(
    predictions_csv,
    index=False,
    encoding="utf-8-sig"
)

# ================================================================
# 12. CREATE EXCEL REPORT
# ================================================================

excel_path = (
    f"{XAI_DIR}/TeluguShield_Integrated_Gradients_Report.xlsx"
)

with pd.ExcelWriter(excel_path, engine="openpyxl") as writer:

    summary_df.to_excel(
        writer,
        sheet_name="XAI_Examples",
        index=False
    )

    attribution_df.to_excel(
        writer,
        sheet_name="Token_Attributions",
        index=False
    )

    predictions_df.to_excel(
        writer,
        sheet_name="Model_Predictions",
        index=False
    )

# ================================================================
# 13. FINAL RESULTS
# ================================================================

print("\n" + "=" * 70)
print("INTEGRATED GRADIENTS COMPLETED")
print("=" * 70)

print("\nXAI EXAMPLE SUMMARY:")
display(summary_df)

print("\nTOKEN ATTRIBUTION SAMPLE:")
display(attribution_df.head(20))

print("\nSAVED FILES:")
print(excel_path)
print(attribution_csv)
print(summary_csv)
print(predictions_csv)

print("\nTOTAL EXPLANATIONS:", len(summary_df))
print("TOTAL TOKEN ATTRIBUTIONS:", len(attribution_df))

print("\nHTML REPORTS:")
for path in summary_df["HTML_Report"].tolist():
    print(path)

print("\nAll results saved in:")
print(XAI_DIR)

print("\nIMPORTANT:")
print("Attributions indicate model contributions relative to the")
print("chosen baseline. They are not proof of linguistic intent.")
```
